1.5 Swap two variables twice: once with a temporary variable, once with tuple
unpacking (
a, b = b, a).

```python
def Swap():
  a = int(input("Enter first number: "))
  b = int(input("Enter second number: "))

  #with temp variable
  temp = a
  a = b
  b = temp
  print("After swapping with temp variable: ", a, b)

  #with tuple unpacking
  a, b = b, a
  print("After swapping with tuple unpacking: ", a, b)
Swap()
```

1.6 Print the
type() of 5,
5.0 ,
"5" ,
True ,
None ,
[5] ,
(5,) ,
{5} and
{"a": 5}
Add a comment explaining each result.


```python
def findType():
  # 5 is a whole number, representing an Integer
  print(type(5))

  # 5.0 is a number with a decimal point, representing a Floating-point number
  print(type(5.0))

  # "5" is enclosed in quotes, representing a String (text)
  print(type("5"))

  # True is a boolean value representing truth/logical state
  print(type(True))

  # None represents the absence of a value or a null value
  print(type(None))

  # [5] is enclosed in square brackets, representing a List
  print(type([5]))

  # (5,) has a trailing comma inside parentheses, representing a Tuple
  print(type((5,)))

  # {5} contains unique elements without keys, representing a Set
  print(type({5}))

  # {"a": 5} contains key-value pairs, representing a Dictionary (dict)
  print(type({"a": 5}))

# Call the function outside of its definition
findType()
```

1.7 Read principal, rate (%) and years. Print simple interest and compound interest
(yearly compounding), rounded to 2 decimals.

```python
def interest():
  principal = float(input("Enter principal amount: "))
  rate = float(input("Enter rate of interest: "))
  years = int(input("Enter number of years: "))
  simpleInterest = (principal * rate * years) / 100
  compoundInterest = principal * (1 + (rate / 100)) ** years - principal
  print("Simple Interest: ", round(simpleInterest, 2))
  print("Compound Interest: ", round(compoundInterest, 2))
interest()
```

1.8 Read a 3-digit number and print the sum of its digits using only // and %. Example:
472 gives 13

```python
def sumOfDigits():
  num = int(input("Enter a 3-digit number: "))
  sum = 0
  while num > 0:
    digit = num % 10
    sum += digit
    num = num / 10
    num = int(num)
  print("Sum of digits: ", sum)
sumOfDigits()
```

1.10 Read a total number of seconds and print it as HH:MM:SS, with leading zeros.
Example: 3725 gives 01:02:05

```python
def timeConv():
  totalSeconds = int(input("Enter total number of seconds: "))
  hours = totalSeconds // 3600
  minutes = (totalSeconds % 3600) // 60
  seconds = totalSeconds % 60
  print(f"{hours:02}:{minutes:02}:{seconds:02}")
timeConv()
```
