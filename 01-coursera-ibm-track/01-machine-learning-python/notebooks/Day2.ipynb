2.1 Read a number and print whether it is even or odd, and whether it is positive,
negative or zero.

```python
def prblm ():
  num = int(input("Enter the Number: "))
  if num == 0:
    print("Number is Zero")
  elif num > 0 and num % 2 == 0:
    print("Number is Positive and Even")
  elif num > 0 and num % 2 != 0:
    print("Number is Positive and Odd")
  elif num < 0 and num % 2 == 0:
    print("Number is Negative and Even")
  else:
    print("Number is Negative and Odd")
prblm()
```

 2.2 Read three numbers and print the largest without using max(). Test with ties, such
as 5 5 3

```python
def maxOfThree():
  # Read input and ensure we have exactly 3 numbers
  try:
    num = list(map(int, input("Enter three numbers separated by spaces: ").split()))
    if len(num) < 3:
      print("Please enter exactly three numbers.")
      return
  except ValueError:
    print("Please enter valid integers.")
    return

  a, b, c = num[0], num[1], num[2]

  # Determine the largest without using max()
  if a >= b and a >= c:
    largest = a
  elif b >= a and b >= c:
    largest = b
  else:
    largest = c

  print(f"{largest} is Greater")

maxOfThree()
```

2.3 Leap year check. A year is a leap year if it is divisible by 4, except years divisible by
100, unless also divisible by 400. Test with 1900, 2000, 2024, 2026.

```python
def leapYear():
  year = int(input("Enter the Year: "))
  if year % 4 == 0 and (year % 100 != 0 or year % 400 == 0):
    print(f"{year} is a Leap Year")
  else:
    print(f"{year} is not a Leap Year")
leapYear()
```

2.4 Read marks from 0 to 100 and print a grade: A for 90 and above, B for 80 to 89, C for
70 to 79, D for 60 to 69, F below 60. Print an error message for marks outside 0 to 100.

```python
def grade():
  marks = int(input("Enter the Marks: "))
  if marks > 100 or marks < 0:
    print("Please enter the valid Marks 0 - 100")
  elif marks >= 90:
    print("Grade A")
  elif marks >= 80:
    print("Grade B")
  elif marks >= 70:
    print("Grade C")
  elif marks >= 60:
    print("Grade D")
  else:
    print("Grade F")
grade()
```

2.5 Read three side lengths. Print whether they form a valid triangle, and if so whether
it is equilateral, isosceles or scalene. Remember that the sum of any two sides must
exceed the third.

```python
def validTriangle():
  a = int(input("Enter the First Side: "))
  b = int(input("Enter the Second Side: "))
  c = int(input("Enter the Third Side: "))

  if a + b <= c or b + c <= a or c + a <= b:
    print("Invalid Triangle")
    return
  if a == b == c:
    print("Equilateral Triangle")
  elif a == b or b == c or a == c:
    print("Isosceles Triangle")
  else:
    print("Scalene Triangle")
validTriangle()
```

2.6 Calculator: read two numbers and an operator (+ - * / // % **) and print the
result. Show a clear message for division by zero and for an unknown operator.

```python
def calc():
  a = int(input("Enter the First Number: "))
  b = int(input("Enter the Second Number: "))
  op = input("Enter the Operator: ")
  if op == "+":
    print(f"{a} + {b} = {a + b}")
  elif op == "-":
    print(f"{a} - {b} = {a - b}")
  elif op == "*":
    print(f"{a} * {b} = {a * b}")
  elif op == "/":
    if a == 0 or b == 0:
      print("Invalid Division")
    else:
      print(f"{a} / {b} = {a / b}")
  elif op == "%":
    print(f"{a} % {b} = {a % b}")
  elif op == "//":
    if a == 0 or b == 0:
      print("Invalid Division")
    else:
      print(f"{a} // {b} = {a // b}")
  elif op == "**":
    print(f"{a} ** {b} = {a ** b}")
  else:
    print("Invalid Operator")
calc()
```

2.7 Electricity bill with slabs: the first 100 units cost 5 each, the next 100 cost 7 each,
anything above 200 costs 10 each. Read the units and print the bill. Test 0, 100, 101,
200, 250.

```python
def electricityBill():
  units = int(input("Enter the Units: "))
  if units <= 100:
    print(f"Bill = {units * 5}")
  elif units <= 200:
    print(f"Bill = {100 * 5 + (units - 100) * 7}")
  elif units > 200:
    print(f"Bill = {units * 10}")
electricityBill()
```

2.8 Ticket price: base price 200. Age under 5 is free, under 18 pays half, 60 and above
pays half, everyone else pays full. Add a 50 weekend surcharge for paid tickets only.
Read the age and a weekend yes/no.

```python
def ticketPrice():
  age = int(input("Enter the Age: "))
  weekend = input("Enter the Weekend (yes/no): ").strip().lower()

  if age < 0:
    print("Enter a valid age.")
    return

  # Determine base price based on age group
  if age < 5:
    price = 0
  elif age < 18:
    price = 100
  elif age >= 60:
    price = 100
  else:
    price = 200

  # Surcharge of 50 only applies to paid tickets on weekends
  if weekend == "yes" and price > 0:
    price += 50

  print(f"Ticket Price = {price}")

ticketPrice()
```

2.9 Read a, b, c of a quadratic equation. Print two real roots, one repeated root, or a
message that the roots are complex, using the discriminant. Handle a = 0 separately

```python
def quadraticEquation():
  a = int(input("Enter the value of a: "))
  b = int(input("Enter the value of b: "))
  c = int(input("Enter the value of c: "))

  if a == 0:
    print("Invalid Quadratic")
  elif (b ** 2) - (4 * a * c) < 0:
    print("Roots are Complex")
  elif (b ** 2) - (4 * a * c) == 0:
    print("Roots are Repeated")
  else:
    print("Roots are Real")
quadraticEquation()
```

2.10 Rock, paper, scissors: read two players' choices, validate them, and print the
winner or a draw. Use as few if branches as you can while keeping it readable

```python
def rockPaperScissors():
  player1 = input("Enter Player 1's Choice (rock, paper, scissors): ").strip().lower()
  player2 = input("Enter Player 2's Choice (rock, paper, scissors): ").strip().lower()

  if player1 not in ["rock", "paper", "scissors"] or player2 not in ["rock", "paper", "scissors"]:
    print("Invalid Choice")
    return

  if player1 == player2:
    print("Draw")
  elif (player1 == "rock" and player2 == "scissors") or \
       (player1 == "paper" and player2 == "rock") or \
       (player1 == "scissors" and player2 == "paper"):
    print("Player 1 Wins")
  else:
    print("Player 2 Wins")

rockPaperScissors()
```
